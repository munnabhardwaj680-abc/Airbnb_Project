```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Airbnb_data.csv")
df.head(5)
```

```python
df.columns
```

# Checking Missing Value

```python
print(df.isnull().sum())
```

```python
df.info()
```

# Handling Missing Value

```python
df['last review'] = pd.to_datetime(df['last review'],errors = 'coerce')
```

```python
df.info()
```

```python
df.fillna({'reviews per month' : 0, 'last review' : df['last review'].min()},inplace = True)
```

```python
df.dropna(subset = ['NAME','host name'] , inplace = True)
```

```python
print(df.isnull().sum())
```

```python
df = df.drop(columns = ['license','house_rules'],errors='ignore')
```

```python
# Remove dollar signs and convert to float
import re
df['price'] = df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee'] = df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.info()
```

# Remove duplicates from data

```python
df.drop_duplicates(inplace=True)
```

##### Descriptive statics

```python
df.describe()
```

```python
df.to_csv('Airbnb_Cleaned_data.csv',index=False)
```

# Visualization 

#### Q1. What is the distribution of listing prices?

```python
plt.figure(figsize = (10,6))
sns.histplot(df['price'],bins = 50 , kde = True , color= 'red')
plt.title("Distrbution of listing price")
plt.xlabel("Price $")
plt.ylabel("Frequency")
plt.show()
```

#### Conclusion :- The histogram shows a fairly even distribution of
#### listing price across different price ranges,indicating no
#### particular concentration of listings in any specific price range.The KDE line helps visualize this even spread more clearly,
#### confirming that the dataset contain listing with a wide variety of prices

### Q2. How are different room types distributed?

```python
plt.figure(figsize = (8,5))
sns.countplot(x="room type",data=df,color='skyblue')
plt.title("Room Type Distribution")
plt.xlabel("room type")
plt.ylabel("count")
plt.show()
```

#### Q3. How are listing distributed across different neighborhoods?

```python
plt.figure(figsize = (12,8))
sns.countplot(y="neighbourhood group",data=df,color='lightgreen',order=df["neighbourhood group"].value_counts().index)
plt.title("Number of Listings by neighborhood Group")
plt.xlabel("Count")
plt.ylabel("Neighborhood Group")
plt.show()
```

#### Q4. What is the relationship between price and room type

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x='room type',
    y='price',
    data=df,
    palette='Set1',
    hue='room type',
    legend=False
)

plt.title('Price vs Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price ($)')
plt.show()
```

#### Q5. How has the number of reviews changes over with time

```python
df['lasr review'] = pd.to_datetime(df['last review'])
reviews_over_time = df.groupby(df['last review'].dt.to_period('M')).size()
plt.figure(figsize=(12,6))
reviews_over_time.plot(kind='line',color='red')
plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.show()
```
